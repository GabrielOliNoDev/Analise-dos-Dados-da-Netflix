💡 Principais Insights de Negócios
1. O churn é um grande desafio para a retenção
A plataforma apresenta uma taxa geral de churn de 50,3% entre as 5.000 contas pesquisadas, indicando que as estratégias de retenção exigem prioridade executiva imediata.

2. Clientes do plano "Basic" apresentam o maior risco de evasão
A categoria "Basic" registra a maior velocidade de churn (~61,8%), em comparação com os grupos de assinantes "Standard" e "Premium", que demonstram fidelidade significativamente maior.

3. O ponto crítico de inatividade de 30 dias
A inatividade é o principal indicador antecedente de churn:
Usuários inativos por 0 a 10 dias apresentam uma taxa de churn de apenas 18,3%.
Quando a inatividade ultrapassa 30 dias, o churn dispara para mais de 72,5%, atingindo um pico de 78,1% no intervalo de 40 a 50 dias.

4. Sensibilidade em relação ao gateway de pagamento
Assinantes que utilizam criptomoedas (~59,7%) e cartões-presente (gift cards) apresentam o maior risco de evasão, enquanto as cobranças automatizadas via cartão de crédito e débito mantêm uma longevidade substancialmente maior.

5. A receita concentra-se nas assinaturas Premium
Apesar de ter um preço mais elevado (US$ 17,99/mês), o plano Premium gera a maior parcela da receita mensal acumulada de assinaturas (mais de 30.457), impulsionada pela alta adesão a esse perfil de plano.

 6. Estabilidade por Gênero e Região Geográfica
A rotatividade (churn) está relativamente distribuída entre as regiões geográficas (48% - 52%), sendo que a Europa apresenta o risco ligeiramente mais elevado (51,7%) e os gêneros Ação/Drama impulsionam o maior volume de audiência.

🎯 Recomendações Estratégicas de Negócio
Gatilhos Automatizados de Reengajamento (Dias 15 a 25):
Implementar notificações push personalizadas e e-mails com curadoria de sugestões ("O que assistir a seguir") antes que o usuário atinja 30 dias de inatividade.

Aumento de Valor no Plano Básico:
Introduzir recursos de personalização leve ou marcos de fidelidade para reduzir a taxa de evasão de 61,8% no plano de entrada.

Incentivo a Pagamentos Recorrentes com Cartão:
Oferecer descontos na cobrança ou meses de teste adicionais para usuários que migrarem de métodos como pré-pago, criptomoedas ou cartões-presente para a cobrança automática via cartão de crédito ou débito.

🏁 Conclusão
Esta análise abrangente demonstra que a intervenção comportamental proativa — especialmente voltada para contas inativas entre o 14º e o 30º dia — representa a oportunidade de maior retorno sobre o investimento (ROI) para a Netflix conter a rotatividade, proteger a receita recorrente e maximizar o valor do ciclo de vida do cliente (LTV).
